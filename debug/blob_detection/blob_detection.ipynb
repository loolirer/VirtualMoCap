# Blob Detection


This notebook is a study about *OpenCV's* blob detection methods using *CoppeliaSim* as a simulator. 

The goal is to make a test environment where detection parameters in software and scene can be changed with ease. For this:

1. Create a 3D scene with highly reflectable spherical markers;
2. Retrieve *CoppeliaSim's* Vision Sensor image;
3. Check if detection results are reasonable.  

Try different marker sizes, lighting conditions and color properties to experiment different results.

---

```python
# Importing modules...
import numpy as np
import cv2

from virtualmocap.plot.viewer2d import Viewer2D
from virtualmocap.plot.viewer3d import Viewer3D

from coppeliasim_zmqremoteapi_client import RemoteAPIClient

# Init client
client = RemoteAPIClient()  # Client object
sim = client.getObject("sim")  # Simulation object
```

# Setting up and Plotting the 3D Scene

For setting up the scene, there will be a set of markers scattered across the scene.

---

```python
# 3D points to be projected
marker_positions = []
for x in np.linspace(-0.75, 0.75, num=5):
    for y in np.linspace(-0.75, 0.75, num=5):
        marker_positions.append(np.array([x, y, 0.1]))

marker_positions = np.array(marker_positions).T

# Create the Scene Viewer
scene = Viewer3D(title="Marker Distribution", size=5)

# Add elements to the scene
scene.add_frame(np.eye(4), "World Frame", axis_size=0.2)
scene.add_points(marker_positions, "Markers", "black")

# Plot scene
scene.figure.show(renderer="notebook_connected")
```

# Getting CoppeliaSim's Image Feed

To retrieve the Vision Sensor's image in *CoppeliaSim*, the following process will be followed:

1. Get raw image with `sim.getVisionSensorImg`;
2. Flip the image;

The resulting image will be similar to the Vision Sensor's view on the application.

---

```python
# When simulation is not running, ZMQ message handling could be a bit
# slow, since the idle loop runs at 8 Hz by default.
# Setting the idle loop to run at full speed for this program
defaultIdleFps = sim.getInt32Param(sim.intparam_idle_fps)
sim.setInt32Param(sim.intparam_idle_fps, 0)

# Simulation begins here
sim.startSimulation()

# Get the vision sensor handle
vision_sensor_handle = sim.getObject(f"/Vision_sensor")

marker_radius = 0.0075
marker_color = [1, 1, 1]
markers_handle = sim.addDrawingObject(
    sim.drawing_spherepts | sim.drawing_painttag,
    marker_radius,
    0.0,
    -1,
    0,
    None,
    None,
    None,
    marker_color,
)

# Add markers to drawing object
for position in marker_positions.T.tolist():
    sim.addDrawingObjectItem(markers_handle, position)

image_gray = None  # Image will be stored here
resolution = None

# Retrieve images
while (t := sim.getSimulationTime()) < 5:
    # Get grayscale image buffer
    buffer, resolution = sim.getVisionSensorImg(
        vision_sensor_handle, 1
    )  # Set second argument to 1 for grayscale, 0 for RGB

    # Convert buffer into single channel image
    image_unflipped = np.frombuffer(buffer, dtype=np.uint8).reshape(
        resolution[1], resolution[0]
    )

    # In CoppeliaSim images are left to right (x-axis), and bottom to top (y-axis)
    # This is consistent with the axes of vision sensors, pointing Z outwards, Y up
    image_gray = cv2.flip(image_unflipped, 0)

    # Wait for images to come up and check if they're not black images
    if image_gray is not None and np.any(image_gray):
        break  # Stop on the first valid image

# Clear drawing object
sim.addDrawingObjectItem(markers_handle, None)

# Restore the original idle loop frequency:
sim.setInt32Param(sim.intparam_idle_fps, defaultIdleFps)

# Simulation ends here
sim.stopSimulation()
```

# Plotting the Vision Sensor's Image 

The retrieved image will be plotted for visualizing the Vision Sensor's view.

---

```python
# Create the image viewer
plot = Viewer2D(
    title="Coppelia's Vision Sensor Image",
    resolution=resolution,
    image=image_gray,
    graphical=True,
)

# Plot image
plot.figure.show(renderer="notebook_connected")
```

# Defining Blobs and Setting Parameters

The information bellow is taken from the Big Vision's LearnOpenCV website and is available [here](https://learnopencv.com/blob-detection-using-opencv-python-c/).

A Blob is defined by a group of connected pixels in an image that share some common property (e.g, grayscale value).

*OpenCV* provides a convenient way to detect and filter blobs based on different characteristics: the SimpleBlobDetector. This class needs a set of tweakable parameters to work in an specified condition. 

For this, we need to instanciate an object of the detector parameters' class and populate it with data that makes sense for our detection. The following work process is listed bellow and how each parameters affects this process:

1. **Thresholding**: Convert the source images to several binary images by thresholding the source image with thresholds starting at `minThreshold`. 

    These thresholds are incremented by `thresholdStep` until `maxThreshold`;

2. **Grouping**: In each binary image, connected white pixels (Binary Blobs) are grouped;
3. **Merging**: The centers of the binary blobs in the binary images are computed, and blobs located closer than `minDistBetweenBlobs` are merged;
4. **Center & Radius Calculation**:  The centers and radii of the newly merged blobs are computed and returned.

Some filters are applied for discarding unwanted blobs. The available filters can be:

- **Color**: set `filterByColor = True`. Set `blobColor = 0` to select darker blobs, and  `blobColor = 255` for lighter blobs;
- **Size**: set `filterByArea = True`. Set values for `minArea` and `maxArea` to filter blobs with a range of your desired pixel count; 
- **Circularity**: set `filterByCircularity = True`. Set appropriate values for `minCircularity` and `maxCircularity`. 
    The circularity $C$ of a shape is defined by:
    $$C = \frac{4\cdot \pi\cdot {Area}}{{Perimeter}^2}$$ 
- **Convexity**: set `filterByConvexity= True`. Set $0 \leq$ `minConvexity` $\leq 1$ and `maxConvexity` $\leq 1$.
- **Inertia Ratio**: set `filterByInertia = True`.Set $0 \leq$ `minInertiaRatio` $\leq 1$ and `maxInertiaRatio` $\leq 1$.



---

```python
params = cv2.SimpleBlobDetector_Params()

# Check if blob is stable in the three filters
params.minRepeatability = 3

# Three threshold filters
params.minThreshold = 50
params.thresholdStep = 50
params.maxThreshold = (
    params.minThreshold + params.thresholdStep * params.minRepeatability
)

# Minimum distance between blobs is 1 pixel
params.minDistBetweenBlobs = 1

# Filter only dark blobs
params.filterByColor = True
params.blobColor = 0

# Filter only blobs with over 2 pixels
params.filterByArea = True
params.minArea = 2

# Do not filter by convexity to allow distorted blobs to be detected
params.filterByConvexity = False
```

# Detecting Blob Centroids

For detection, a SimpleBlobDetector object will be created using the previous parameters. The `detect` class method will be called to a thresholded image and will return the keypoints that will be converted in a 2D point cloud.

---

```python
# Instanciate marker detector object
marker_detector = cv2.SimpleBlobDetector_create(params)

# Apply threshold to image
thresh = 127
_, image_thresh = cv2.threshold(image_gray, thresh, 255, cv2.THRESH_BINARY_INV)

# Optimization: finding a smaller sub-image that contains all blobs
zero_pixels = np.array(np.where(image_thresh == 0))

# Image blacked out after thresh
if not zero_pixels.size:
    detected_blobs = np.array([])

# Sub-image new corners
margin = 5
u_min, u_max = min(zero_pixels[1]) - margin, max(zero_pixels[1]) + margin
v_min, v_max = min(zero_pixels[0]) - margin, max(zero_pixels[0]) + margin

# Clip values to not go beyond the image limits
u_min, u_max = np.clip([u_min, u_max], 0, image_thresh.shape[1] - 1)
v_min, v_max = np.clip([v_min, v_max], 0, image_thresh.shape[0] - 1)

# Sub-image reference
sub_image_origin = np.array([u_min, v_min])

# Detect keypoints in sub-image
keypoints = marker_detector.detect(
    image_thresh[v_min : v_max + 1, u_min : u_max + 1]
)  # Slice ends are exclusive!

# No valid blob found!
if not keypoints:
    detected_blobs = np.array([])

else:
    # If valid blobs were found, make detected blobs matrix in the original image reference
    detected_blobs = np.array([k.pt for k in keypoints]) + sub_image_origin
```

# Plotting Detected Blob Centroids

The detected blob centroids will be plotted over the image to check if the result is reasonable. 

---

```python
if not detected_blobs.size:
    print("No blob detected!")

else:
    plot.add_points(points=detected_blobs.T, name="Detected Blobs", color="White")

    # Plot image
    plot.figure.show(renderer="notebook_connected")
```
