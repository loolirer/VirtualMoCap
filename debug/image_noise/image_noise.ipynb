# Image Noise Model

This notebook is a study about the imaging noise models using *CoppeliaSim* as a simulator to generate synthetic data in a virtual environment. 

To simulate the noise of a real image sensor in a synthetic image generated by *CoppeliaSim's* render. For this:
1. Instanciate a Vision Sensor in *CoppeliaSim*;
2. Get ground truth image;
3. Apply noise to the image in order to match an arbitrarily chosen ${SNR}$;

The goal is to produce an image with the chosen ${SNR}$. 

---

```python
# Importing modules...
import numpy as np

import sys

sys.path.append("../..")  # Go back to base directory

from modules.plot.viewer2d import Viewer2D

from modules.integration.coppeliasim.camera import CoppeliaSim_Camera

from coppeliasim_zmqremoteapi_client import RemoteAPIClient

# Init client
client = RemoteAPIClient()  # Client object
sim = client.getObject("sim")  # Simulation object
```

# Instanciating Camera Object

To generate the data regarding the mathematical model of the vision sensor in *CoppeliaSim*, it will be instanciated a Camera Obejct that matches the Vision Sensor's intrinsic and extrinsic parameters. 

---

```python
# Get the vision sensor handle
vision_sensor_handle = sim.getObject(f"/Vision_sensor")

# Get Coppelia's camera pose
pose = np.vstack(
    (
        np.array(sim.getObjectMatrix(vision_sensor_handle)).reshape((3, 4)),
        np.array([0, 0, 0, 1]),
    )
)

# X and Y axis of Coppelia's Vision Sensor are inverted
pose[:, :2] *= -1  # Multiplies by -1 the first two columns

camera = CoppeliaSim_Camera(
    vision_sensor_handle=vision_sensor_handle,
    resolution=(960, 720),
    fov_degrees=60.0,
    pose=pose,
)
```

# Getting CoppeliaSim's Image Feed

Retrieving VisionSensor groundtruth image with `sim.getVisionSensorImg`.

---

```python
# When simulation is not running, ZMQ message handling could be a bit
# slow, since the idle loop runs at 8 Hz by default.
# Setting the idle loop to run at full speed for this program
defaultIdleFps = sim.getInt32Param(sim.intparam_idle_fps)
sim.setInt32Param(sim.intparam_idle_fps, 0)

# Simulation begins here
sim.startSimulation()

image_noiseless = None

while (t := sim.getSimulationTime()) < 5:
    image_noiseless = camera.get_image(sim.getVisionSensorImg)

    # Wait for images to come up and check if they're not black images
    if image_noiseless is not None and np.any(image_noiseless):
        break  # Stop on the first valid image

# Restore the original idle loop frequency:
sim.setInt32Param(sim.intparam_idle_fps, defaultIdleFps)

# Simulation ends here
sim.stopSimulation()
```

# Plotting the Vision Sensor's Image 

The Vision Sensor's retrieved image will be plotted for reference. It's important to note that, since *CoppeliaSim* does not have any noise model, the image retrieved from the Vision Sensor is the ground truth signal.

---

```python
# Create the image viewer
plot = Viewer2D(
    title="Noiseless Image",
    resolution=camera.resolution,
    image=image_noiseless,
    graphical=True,
)

# Plot image
plot.figure.show(renderer="notebook_connected")
```

# Defining Signal to Noise Ratio of an Image

The imaging noise is composed of random spikes of luminance change in places where it is expected to be uniform or maybe changing gradually. If the change of luminance affects equally the RGB components of a pixel, then it is classified as *luminance noise*. Since the image is single-channel, this will be the modelled noise.

The Signal to Noise Ratio $({SNR})$ is a measure that compares the level of a desired signal to the level of background noise. In the case of images, it referes to "how much of the image" is there in relation to the noise that affects it. 

Mathmatically, the Signal to Noise Ratio is defined as such:

$${SNR} = \frac{\mu}{\sigma}$$

Where $\mu$ is the average value of the signal, and $\sigma$ is its respective standard deviation.

In the case of an image single-channel image, this concept should only apply separately to evenly lit portions of it, since the noise is being measured in uniform lighting. In this case:

$${SNR} = \frac{\mu_p}{\sigma_p}$$

Where $\mu_p$ is the average luminosity of the pixels, and $\sigma_p$ is their respective standard deviation.

Since the ${SNR}$ is usually represented in decibels (dB), the conversion is done by:

$${SNR}_{dB} = 20 \cdot \log_{10}{SNR}$$

For a synthetic image in whole uniform lighting, where there is no deviation in pixel by pixel luminosity, $\sigma_p = 0$ and therefore the ${SNR}$ should be as high as it can get.

---

```python
def SNR(signal, unit="dB"):
    mu = np.mean(signal)  # Average value of the signal
    sigma = np.std(signal)  # Standard deviation of the signal

    if sigma == 0:
        return np.inf  # Returns infinity in case the signal has no noise
    else:
        snr = mu / sigma

        if unit == "ratio":
            return snr

        elif unit == "dB":
            return 20 * np.log10(snr)  # SNR in decibels
```

# Simulating Signal to Noise Ratio

To simulate an arbitrary Signal to Noise Ratio in a synthetic image the following steps bellow will suffice:
1. Assume a pixel's value is the average signal value (since we're analysing the ground truth);
2. Create a standard deviation matrix of the image size following: $\sigma_p = \mu_p / {SNR}$;
3. Generate new image based on the average matrix of the image (itself) and the standard deviation matrix of the image;

According to the [DxOMark](https://www.dxomark.com), the following table sets a reference for what's considered an image with adequate and poor quality by its ${SNR}$:

| Noise Category | dB |
|----------------|----|
| Good           | $>38$ |
| Acceptable     | $>32$ & $<38$ |
| Poor Quality   | $>20$ & $<32$ |
| Non acceptable | $<20$ |

---

```python
snr_dB = 20  # Desired SNR value in db
snr = np.power(10, (snr_dB / 20))  # Converted for ratio

# Apply noise
image_noisy = None

if snr == np.inf:
    image_noisy = image_noiseless

else:
    sigma = image_noiseless / snr  # Standard deviation matrix of the image
    mu = image_noiseless.astype(np.float32)  # Average signal of each pixel in the image

    image_noisy = np.random.normal(
        mu, sigma
    )  # Generate noisy image based on gaussian noise (normal distribution)

    # Clip negative pixel values to 0 and saturated pixel values a maximum 8-bit interger value
    image_noisy = np.clip(0, 255, image_noisy).astype(np.uint8)
```

# Plotting the Noisy Image

The noised image will be plotted for visualizing the noise profile and the different effects of different Signal to Noise Ratios.

---

```python
# Create the image viewer
plot_noised = Viewer2D(
    title=f"Noisy Image with {snr_dB} dB SNR",
    resolution=camera.resolution,
    image=image_noisy,
    graphical=True,
)

# Plot image
plot_noised.figure.show(renderer="notebook_connected")
```
